```python
#These classes provide a structured way to represent COBOL program elements in Python before converting them into Java
from dataclasses import dataclass, field
from typing import List

#Base class for all AST node types.
@dataclass
class ASTNode:
    pass

#Stores a COBOL DISPLAY statement.
@dataclass
class DisplayNode(ASTNode):
    raw_text: str
    line: int

#Stores a COBOL MOVE statement.
@dataclass
class MoveNode(ASTNode):
    raw_text: str
    line: int

#Stores statements that your translator hasn't classified specifically.
@dataclass
class OtherStatementNode(ASTNode):
    raw_text: str
    line: int

#Represents the program and holds its statements.
@dataclass
class ProgramNode(ASTNode):
    name: str
    statements: List[ASTNode] = field(default_factory=list)

print("All AST classes defined successfully.")
```

```python
# Build a conservative statement inventory from the selected source.
# This deliberately preserves unsupported statements for reporting instead of
# pretending that a partial translation is complete.
import re
from Cobol85Visitor import Cobol85Visitor

with open(selected_file, "r", encoding="utf-8", errors="replace") as f:
    source_lines = f.readlines()

# A lightweight source-level collector is used here because parser contexts can
# span whole paragraphs and ctx.getText() removes whitespace. It is not a full
# COBOL parser; multi-line statements are joined when continuation is obvious.
VERBS = (
    "DISPLAY", "MOVE", "PERFORM", "SEARCH", "IF", "ELSE", "END-IF",
    "ACCEPT", "COMPUTE", "ADD", "SUBTRACT", "MULTIPLY", "DIVIDE",
    "CONTINUE", "EXIT", "OPEN", "READ", "WRITE", "REWRITE", "CLOSE",
    "INITIALIZE", "STRING", "UNSTRING", "INSPECT", "SET", "GO", "STOP",
    "CALL", "EVALUATE", "END-PERFORM", "END-READ", "END-WRITE"
)
verb_pattern = re.compile(r"^\s*(" + "|".join(re.escape(v) for v in sorted(VERBS, key=len, reverse=True)) + r")\b", re.I)

def clean_cobol_line(line):
    # Fixed-format COBOL comments usually have * or / in column 7.
    if len(line) >= 7 and line[6] in "*/":
        return ""
    # Strip inline comments only when they begin with *>.
    return re.split(r"\*>", line.rstrip("\n"), maxsplit=1)[0].strip()

statements = []
inside_procedure = False
buffer = ""
buffer_line = 0
for line_no, original in enumerate(source_lines, start=1):
    line = clean_cobol_line(original)
    if not line:
        continue
    upper = line.upper()
    if "PROCEDURE DIVISION" in upper:
        inside_procedure = True
        continue
    if not inside_procedure:
        continue
    # Paragraph/section labels are not executable statements.
    if re.match(r"^[A-Z0-9-]+(?:\s+SECTION)?\.$", upper):
        if buffer:
            statements.append((buffer_line, buffer.strip()))
            buffer, buffer_line = "", 0
        continue
    if verb_pattern.match(line):
        if buffer:
            statements.append((buffer_line, buffer.strip()))
        buffer, buffer_line = line, line_no
    elif buffer:
        buffer += " " + line
    # If line ends with a period, treat it as the end of a sentence.
    if buffer and line.rstrip().endswith("."):
        statements.append((buffer_line, buffer.strip()))
        buffer, buffer_line = "", 0
if buffer:
    statements.append((buffer_line, buffer.strip()))

builder_statements = []
for line_no, raw in statements:
    keyword_match = verb_pattern.match(raw)
    keyword = keyword_match.group(1).upper() if keyword_match else ""
    if keyword == "DISPLAY":
        builder_statements.append(DisplayNode(raw, line_no))
    elif keyword == "MOVE":
        builder_statements.append(MoveNode(raw, line_no))
    else:
        builder_statements.append(OtherStatementNode(raw, line_no))

program_match = re.search(r"PROGRAM-ID\s*\.\s*([A-Z0-9-]+)", cobol_source, re.I)
program_name = program_match.group(1) if program_match else os.path.splitext(os.path.basename(selected_file))[0]
program_ast = ProgramNode(name=program_name, statements=builder_statements)

print("Program:", program_ast.name)
print("Procedure statements inventoried:", len(program_ast.statements))
print("Note: this source-level inventory is conservative, not a complete COBOL grammar/semantic AST.")
```

```python
#translate COBOL DISPLAY statements into equivalent Java System.out.println() statements
import re

def to_java_identifier(name):
    parts = name.lower().split("-")
    ident = parts[0] + "".join(part.capitalize() for part in parts[1:])
    if ident in {"class", "public", "static", "void", "int", "long", "double", "float", "boolean", "new", "return"}:
        ident += "Value"
    return ident

def java_string(value):
    value = value.replace("\\", "\\\\").replace('"', '\\"').replace("\n", "\\n").replace("\r", "\\r")
    return '"' + value + '"'

def translate_display(node):
    text = node.raw_text.strip().rstrip(".")
    match = re.match(r"DISPLAY\s+(.+)$", text, re.I | re.S)
    if not match:
        return None
    content = match.group(1).strip()
    # Support simple DISPLAY operands only. Complex COBOL formatting is reported unsupported.
    tokens = re.findall(r"'(?:[^']|'')*'|\"(?:[^\"]|\"\")*\"|[A-Za-z][A-Za-z0-9-]*|\d+(?:\.\d+)?", content)
    residue = re.sub(r"'(?:[^']|'')*'|\"(?:[^\"]|\"\")*\"|[A-Za-z][A-Za-z0-9-]*|\d+(?:\.\d+)?", "", content)
    if not tokens or residue.strip():
        return None
    parts = []
    for token in tokens:
        if token.startswith("'") and token.endswith("'"):
            parts.append(java_string(token[1:-1].replace("''", "'")))
        elif token.startswith('"') and token.endswith('"'):
            parts.append(java_string(token[1:-1].replace('""', '"')))
        elif re.fullmatch(r"\d+(?:\.\d+)?", token):
            parts.append(token)
        else:
            parts.append(to_java_identifier(token))
    return "System.out.println(" + " + ".join(parts) + ");"
```

```python
#convert supported MOVE statements into Java assignments in your COBOL-to-Java migration assistant.
import re

def translate_move(node):
    text = node.raw_text.strip().rstrip(".").strip()
    atom = r"(?:'(?:[^']|'')*'|\"(?:[^\"]|\"\")*\"|[A-Za-z][A-Za-z0-9-]*|[-+]?\d+(?:\.\d+)?)"
    # MOVE literal-or-identifier TO TABLE(SUBSCRIPT)
    indexed = re.fullmatch(r"MOVE\s+(" + atom + r")\s+TO\s+([A-Za-z][A-Za-z0-9-]*)\s*\(\s*([A-Za-z][A-Za-z0-9-]*)\s*\)", text, re.I)
    simple = re.fullmatch(r"MOVE\s+(" + atom + r")\s+TO\s+([A-Za-z][A-Za-z0-9-]*)", text, re.I)
    match = indexed or simple
    if not match:
        return None
    value, target = match.group(1), match.group(2)
    if value.startswith("'") and value.endswith("'"):
        java_value = java_string(value[1:-1].replace("''", "'"))
    elif value.startswith('"') and value.endswith('"'):
        java_value = java_string(value[1:-1].replace('""', '"'))
    elif re.fullmatch(r"[-+]?\d+(?:\.\d+)?", value):
        java_value = value
    else:
        java_value = to_java_identifier(value)
    java_target = to_java_identifier(target)
    if indexed:
        index = to_java_identifier(match.group(3))
        # This mapping assumes a one-based COBOL numeric subscript and a Java array.
        return f"{java_target}[{index} - 1] = {java_value};"
    return f"{java_target} = {java_value};"
```

```python
# Translate supported statements and retain a complete unsupported inventory.
generated_statements = []
unsupported_statements = []
for node in program_ast.statements:
    java_code = None
    if isinstance(node, DisplayNode):
        java_code = translate_display(node)
    elif isinstance(node, MoveNode):
        java_code = translate_move(node)
    if java_code:
        generated_statements.append(java_code)
    else:
        unsupported_statements.append(node)

print("Generated Java statements:")
for statement in generated_statements:
    print("  ", statement)
print("\nUnsupported statements (must be implemented before claiming a complete migration):")
for node in unsupported_statements:
    print(f"Line {node.line}: {node.raw_text}")
print(f"\nTranslated {len(generated_statements)} of {len(program_ast.statements)} inventoried statements.")
```

```python
#creates a Java implementation of CPSEQVR.CBL and saves it as /content/CPSEQVR.java
from pathlib import Path

java_code = r'''
import java.io.BufferedReader;
import java.io.BufferedWriter;
import java.io.IOException;
import java.nio.charset.StandardCharsets;
import java.nio.file.Files;
import java.nio.file.Path;
import java.nio.file.Paths;
import java.nio.file.StandardOpenOption;
import java.util.Arrays;

public class CPSEQVR {

    private static final int FIELD_LENGTH = 5;
    private static final int MAX_FIELDS = 10;
    private static final int MAX_RECORD_LENGTH = 50;

    public static void main(String[] args) {
        Path inputPath = Paths.get(
            args.length > 0 ? args[0] : "VARFILE1"
        );
        Path outputPath = Paths.get(
            args.length > 1 ? args[1] : "VARFILE2"
        );

        int recordCount = 0;

        try (BufferedReader reader = Files.newBufferedReader(
                 inputPath, StandardCharsets.UTF_8)) {

            // Open output only after input was successfully opened,
            // matching the sequence of the COBOL program.
            try (BufferedWriter writer = Files.newBufferedWriter(
                     outputPath,
                     StandardCharsets.UTF_8,
                     StandardOpenOption.CREATE,
                     StandardOpenOption.TRUNCATE_EXISTING,
                     StandardOpenOption.WRITE)) {

                String inputRecord;

                while ((inputRecord = reader.readLine()) != null) {
                    int inputLength = inputRecord.length();

                    if (inputLength < FIELD_LENGTH
                            || inputLength > MAX_RECORD_LENGTH) {
                        throw new IOException(
                            "Input record length outside 5..50: "
                            + inputLength
                        );
                    }

                    if (inputLength % FIELD_LENGTH != 0) {
                        throw new IOException(
                            "Input record length must be a multiple of 5: "
                            + inputLength
                        );
                    }

                    // COMPUTE FIELD-COUNT =
                    //     IN-RECLEN / LENGTH OF RECORD-FIELD
                    int fieldCount = inputLength / FIELD_LENGTH;

                    // The program adds one more five-character field.
                    fieldCount++;

                    if (fieldCount > MAX_FIELDS) {
                        throw new IOException(
                            "Cannot append XXXXX: RECORD-FIELD "
                            + "allows at most 10 fields."
                        );
                    }

                    // MOVE INPUT-RECORD TO RECORD-AREA
                    char[] recordArea = new char[MAX_RECORD_LENGTH];
                    Arrays.fill(recordArea, ' ');
                    inputRecord.getChars(
                        0, inputLength, recordArea, 0
                    );

                    // MOVE 'XXXXX' TO RECORD-FIELD(FIELD-COUNT)
                    int start = (fieldCount - 1) * FIELD_LENGTH;
                    Arrays.fill(
                        recordArea,
                        start,
                        start + FIELD_LENGTH,
                        'X'
                    );

                    // COMPUTE OUT-RECLEN =
                    //     FIELD-COUNT * LENGTH OF RECORD-FIELD
                    int outputLength = fieldCount * FIELD_LENGTH;

                    // WRITE OUTPUT-RECORD FROM RECORD-AREA
                    writer.write(recordArea, 0, outputLength);
                    writer.newLine();

                    // ADD 1 TO RECORD-COUNT
                    recordCount++;
                }
            }

            System.out.println(
                "NUMBER OF RECORDS PROCESSED: " + recordCount
            );

        } catch (IOException e) {
            System.err.println(
                "File processing failed: " + e.getMessage()
            );
            System.exit(1);
        }
    }
}
'''

java_path = Path("/content/CPSEQVR.java")
java_path.write_text(java_code, encoding="utf-8")

print(f"Generated Java file: {java_path}")
print(java_code)
```

```python
# Extract simple COBOL data declarations from the selected program.
# This is an inventory aid, not yet a complete PIC/OCCURS/file-record translator.
import re
cobol_lines = cobol_source.splitlines()
data_items = []
inside_data = False
for line_number, line in enumerate(cobol_lines, start=1):
    if len(line) >= 7 and line[6] in "*/":
        continue
    stripped = re.split(r"\*>", line, maxsplit=1)[0].strip()
    upper = stripped.upper()
    if "DATA DIVISION" in upper:
        inside_data = True
        continue
    if "PROCEDURE DIVISION" in upper:
        break
    if not inside_data or not stripped:
        continue
    m = re.match(r"^(\d{1,2})\s+([A-Z0-9-]+)\s*(.*)$", stripped, re.I)
    if not m:
        continue
    level, name, clauses = int(m.group(1)), m.group(2).upper(), m.group(3).strip()
    pic = re.search(r"\bPIC(?:TURE)?\s+(?:IS\s+)?([A-Z0-9()VXS9P.+-]+)", clauses, re.I)
    occurs = re.search(r"\bOCCURS\s+(\d+)(?:\s+TO\s+(\d+))?", clauses, re.I)
    data_items.append({"line": line_number, "level": level, "name": name,
                       "pic": pic.group(1).upper() if pic else None,
                       "occurs": occurs.group(0).upper() if occurs else None,
                       "clauses": clauses})
print("Data items found:", len(data_items))
for item in data_items:
    print(item)
```

```python
# Summarize declarations that need semantic mapping before Java generation.
if not data_items:
    print("No simple data declarations were detected. Check source format and DATA DIVISION.")
else:
    print("Declaration summary:")
    for item in data_items:
        if item["pic"] is None:
            print(f"Line {item['line']}: {item['name']} has no simple PIC clause (group/file item or unsupported declaration).")
        if item["occurs"]:
            print(f"Line {item['line']}: {item['name']} is a table ({item['occurs']}); Java array size/type must be mapped.")
```

```python
import os
import warnings
warnings.filterwarnings("ignore")

import instructor
import google.generativeai as genai
from pydantic import BaseModel
from deepeval.models import DeepEvalBaseLLM
from google.colab import userdata

GOOGLE_API_KEY = userdata.get('Gemini_API_Key') # os.getenvrion('GOOGLE_API_KEY')

genai.configure(api_key=GOOGLE_API_KEY)

class JudgeResponse(BaseModel):
    score: int
    reason: str


class GeminiJudge(DeepEvalBaseLLM):
    """
    Wraps Google Gemini 2.5 Flash as a DeepEval judge LLM.

    DeepEval passes a Pydantic `schema` to generate() so the judge can return
    structured JSON scores. This implementation uses the `instructor` library
    with instructor.from_gemini() — the pattern from the official DeepEval docs.

    All DeepEval metrics accept a `model=` parameter:
        GEval(..., model=gemini_judge)
    """

    def __init__(self, model_name: str = "gemini-3.1-flash-lite"):
        self.model_name = model_name
        self._client = genai.GenerativeModel(model_name)

    def load_model(self):
        return self._client

    def get_model_name(self) -> str:
        return self.model_name

    def generate(self, prompt: str, schema: BaseModel) -> BaseModel:
        client = self.load_model()
        instructor_client = instructor.from_gemini(
            client=client,
            mode=instructor.Mode.GEMINI_JSON,
        )
        return instructor_client.messages.create(
            messages=[{"role": "user", "content": prompt}],
            response_model=schema,
        )

    async def a_generate(self, prompt: str, schema: BaseModel) -> BaseModel:
        # Async variant — delegates to sync; swap for async Gemini client if needed
        return self.generate(prompt, schema)




# Instantiate once — reuse across all metrics
gemini_judge = GeminiJudge(model_name="gemini-3.6-flash")
# gemini_judge = GeminiJudge(model_name="gemini-2.5-flash", JudgeSchema())

print(f"Judge LLM  : {gemini_judge.get_model_name()}")
print("Pass  model=gemini_judge  to any DeepEval metric to use gemini-3.6-flash as the judge.")
```

```python

from deepeval.metrics import GEval
from deepeval.test_case import LLMTestCaseParams

# ── G-Eval: Transpilation Correctness ────────────────────────────────────────
correctness_metric = GEval(
    name="Transpilation Correctness",
    criteria=(
        "Determine whether the generated Java code preserves the behavior "
        "of the original COBOL program. Check assignments, data types, "
        "arithmetic calculations, record processing, and output behavior. "
        "Penalize incorrect logic and behavioral differences."
    ),
    evaluation_params=[
        LLMTestCaseParams.INPUT,
        LLMTestCaseParams.ACTUAL_OUTPUT,
        LLMTestCaseParams.EXPECTED_OUTPUT,
    ],
    model=gemini_judge,
    threshold=0.9,
)

# ── G-Eval: Code Completeness ────────────────────────────────────────────────
completeness_metric = GEval(
    name="Code Completeness",
    criteria=(
        "Evaluate whether the generated Java code contains all essential "
        "components required by the original COBOL program, including variable "
        "declarations, file input and output, loop and EOF handling, conditional "
        "logic, arithmetic operations, record writing, resource closing, and "
        "error handling. Penalize missing statements, undeclared variables, "
        "incomplete control flow, and code that cannot compile."
    ),
    evaluation_params=[
        LLMTestCaseParams.INPUT,
        LLMTestCaseParams.ACTUAL_OUTPUT,
        LLMTestCaseParams.EXPECTED_OUTPUT,
    ],
    model=gemini_judge,
    threshold=0.9,
)

# ── G-Eval: Instruction Following ────────────────────────────────────────────
instruction_following_metric = GEval(
    name="Instruction Following",
    criteria=(
        "Determine whether the transpiler followed the migration requirements "
        "and generated Java code corresponding to the provided COBOL source. "
        "Check whether essential logic is preserved, unsupported statements "
        "are identified, and the generated output matches the requested format."
    ),
    evaluation_params=[
        LLMTestCaseParams.INPUT,
        LLMTestCaseParams.ACTUAL_OUTPUT,
        LLMTestCaseParams.EXPECTED_OUTPUT,
    ],
    model=gemini_judge,
    threshold=0.7,
)

print("G-Eval metrics defined (judge: Gemini):")
for m in [
    correctness_metric,
    completeness_metric,
    instruction_following_metric,
]:
    print(f"  [{m.name}]  threshold={m.threshold}")
```

```python
for metric in [
    correctness_metric,
    completeness_metric,
    instruction_following_metric,
]:
    metric.measure(test_case)

    print(f"\nMetric: {metric.name}")
    print(f"Score: {metric.score}")
    print(f"Reason: {metric.reason}")
    print(f"Threshold: {metric.threshold}")
    print(f"Passed: {metric.is_successful()}")
```

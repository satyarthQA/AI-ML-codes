````python
# downloads the official California city/place boundaries from the U.S. Census Bureau and loads them into GeoPandas.

import requests
import geopandas as gpd
import zipfile
from io import BytesIO

STATE = "California"

STATE_FIPS = {
    "California": "06",
}

state_fips = STATE_FIPS[STATE]

# Census California places
url = (
    f"https://www2.census.gov/geo/tiger/TIGER2025/"
    f"PLACE/tl_2025_{state_fips}_place.zip"
)

print("Downloading:", url)

response = requests.get(url)
response.raise_for_status()

# Read ZIP
with zipfile.ZipFile(BytesIO(response.content)) as z:

    z.extractall("/tmp/california_places")

# Find shapefile
import os

shp_files = [
    f for f in os.listdir("/tmp/california_places")
    if f.endswith(".shp")
]

shp_path = "/tmp/california_places/" + shp_files[0]

# Create cities GeoDataFrame
cities = gpd.read_file(shp_path)

print("Cities loaded:", len(cities))
print(cities.head())
````

````python
#To link different temperature to California's city.
#Here we have set city name as Los Angeles and linked this city to first 3 tiles of API response.

import pandas as pd
features = data["data"]["result"]["map_data"]["features"]

all_temperature_data = []

city_name = "Los Angeles"
start = int(input("Enter start tile: "))
end = int(input("Enter end tile: "))

for feature in features[start:end]:
    properties = feature["properties"]

    all_temperature_data.append({
        "city": city_name,
        "tile_id": properties["tile_id"],
        "average_temperature": properties["average_temperature"],
        "min_temperature": properties["min_temperature"],
        "max_temperature": properties["max_temperature"]
    })

df = pd.DataFrame(all_temperature_data)

print(df)
````

````python
#Combine all the FortyGuard temperature tiles belonging to each city and creates a city-level temperature summary

city_summary = (
    df.groupby("city")
      .agg(
          average_temperature=("average_temperature", "mean"),
          minimum_temperature=("min_temperature", "min"),
          maximum_temperature=("max_temperature", "max"),
          tile_count=("tile_id", "count")
      )
      .reset_index()
)

print(city_summary)
````

````python
# Set temperature constraint

def get_heat_risk(avg_temp):

    if avg_temp > 35:
        return "VERY HIGH"

    elif avg_temp >= 32:
        return "HIGH"

    elif avg_temp >= 28:
        return "MEDIUM"

    else:
        return "LOW"
````

````python
#For checking heat risk of city

df["heat_risk"] = df.apply(
    lambda row: get_heat_risk(
        row["average_temperature"]
    ),
    axis=1
)
print(df)
````
